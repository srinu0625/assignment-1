```python
start_date = (datetime.today() - timedelta(days=730)).strftime('%Y-%m-%d')
end_date = datetime.today().strftime('%Y-%m-%d')

# RICs for each commodity/index
ric_map = {
    'Gasoil': 'LGOc1',
    'HeatingOilIndex': '.BCOMHOC',
    'WTI': 'CLc1',
    'Brent': 'LCOc1',
    'RBOB_2nd': 'RBc2',
}

# Pull data for each instrument
all_data = {}

for label, ric in ric_map.items():
    try:
        df = ek.get_timeseries(
            ric,
            start_date=start_date,
            end_date=end_date,
            interval='minute',
            fields='CLOSE'
        )
        df.rename(columns={'CLOSE': label}, inplace=True)
        all_data[label] = df[label]
        print(f"✔ Data fetched for {label} ({ric})")
    except Exception as e:
        print(f"❌ Error fetching {label} ({ric}): {e}")

# Merge into one DataFrame
combined_df = pd.concat(all_data.values(), axis=1)

# Drop rows where all values are missing
combined_df.dropna(how='all', inplace=True)
df=combined_df
# Preview
print(combined_df.tail())
```

```python
asset_map = {
    'CLc1': 'WTI',
    'LCOc1': 'Brent',
    '.BCOMHOC': 'HeatingOilIndex',
    'RBOBc2': 'RBOB_2nd',
    'LGOc1': 'Gasoil'
}

tickers = list(asset_map.keys())
labels = list(asset_map.values())

```

```python
def update_data():
    global df
    while True:
        try:
            data, _ = ek.get_data(tickers, ['TR.LastPrice'])

            new_row = {}
            now = pd.Timestamp.now()
            for ric, label in asset_map.items():
                try:
                    price = data.loc[data.index == ric, 'TR.LASTPRICE'].values[0]
                    if pd.isna(price):
                        price = df[label].iloc[-1]
                except:
                    price = df[label].iloc[-1] if label in df.columns and len(df) > 0 else None

                new_row[label] = price

            new_df = pd.DataFrame([new_row], index=[now])
            df = pd.concat([df, new_df]).tail(1000)
            print(f"[{now:%Y-%m-%d %H:%M:%S}] ✅ Row added | df shape: {df.shape}")

        except Exception as e:
            print(f"[{datetime.now():%Y-%m-%d %H:%M:%S}] ❌ Error: {e}")

        time.sleep(60)  # ⬅️ This MUST be here
```

```python
# Start the thread (once per kernel session)
import threading
threading.Thread(target=update_data, daemon=True).start()
print("🔁 Background thread started.")
```

```python
asset_pairs = list(combinations(df.columns, 2))

significant_pairs = []

for a1, a2 in asset_pairs:
    score, pvalue, _ = coint(df[a1], df[a2])
    if pvalue < 0.3:
        significant_pairs.append((f"{a1} ~ {a2}", pvalue, score))
    print(a1,a2,pvalue)

# Convert and print
sig_df = pd.DataFrame(significant_pairs, columns=["Pair", "pvalue", "statistic"])
sig_df.sort_values(by="pvalue", inplace=True)

print("\n✅ Significantly Cointegrated Pairs (p < 0.05):")
print(sig_df)
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

pairs = list(df.columns)

for i in range(len(pairs)):
    for j in range(i + 1, len(pairs)):  # Avoid x = y
        x_name = pairs[i]
        y_name = pairs[j]
        
        x = df[x_name].astype(float)
        y = df[y_name].astype(float)

        # Add constant term
        x_const = sm.add_constant(x)

        # Fit regression: y = alpha + beta * x
        model = sm.OLS(y, x_const).fit()
        alpha = model.params['const']
        beta = model.params[x_name]

        # Predicted y and spread
        y_pred = alpha + beta * x
        spread = y - y_pred

        # Plot actual vs fitted
        plt.figure(figsize=(14, 6))
        plt.plot(df.index, y, label=f'{y_name} (Actual)', color='blue')
        plt.plot(df.index, y_pred, label=f'Fitted: {alpha:.2f} + {beta:.2f} × {x_name}', color='orange')
        plt.legend()
        plt.title(f'Regression: {y_name} ~ {x_name}')
        plt.grid(True)
        plt.show()

        # Plot spread
        plt.figure(figsize=(14, 4))
        plt.plot(df.index, spread, label='Spread (Residual)', color='green')
        plt.axhline(0, color='black', linestyle='--')
        plt.legend()
        plt.title(f'Spread: {y_name} - Fitted({x_name})')
        plt.grid(True)
        plt.show()
```
